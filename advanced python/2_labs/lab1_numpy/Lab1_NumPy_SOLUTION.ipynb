# Lab 1 — Daily Sales Analysis with NumPy — **SOLUTION**
A small shop wants to know: how much did we sell, what was the average, which day was best, which days were above a target?

## Dataset

```python
import numpy as np

days = np.array(["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"])
sales = np.array([120, 150, 95, 180, 160, 210, 175])
```

## Exercise 1 — Create and inspect

```python
print(sales)
print("First value:", sales[0])      # 120
print("Last value :", sales[-1])     # 175
print("shape:", sales.shape)         # (7,)  -> a 1D array of 7 values
print("size :", sales.size)          # 7
print("Tuesday -> Friday:", sales[1:5])   # indices 1,2,3,4 (5 excluded) -> [150  95 180 160]
```

## Exercise 2 — Weekly performance (no loops!)

```python
total = sales.sum()
average = sales.mean()
print("Total  :", total)               # 1090
print("Average:", round(average, 2))   # 155.71
print("Min/Max:", sales.min(), sales.max())   # 95 210
best_index = sales.argmax()            # position of the max -> 5
print("Best day index:", best_index, "->", days[best_index])   # Saturday
```

**Answers:** total weekly sales = **1090**; average daily sales ≈ **155.71**; best day = **Saturday** (210).

## Exercise 3 — Filtering with boolean masks

```python
mask_150 = sales > 150
print(mask_150)                       # [False False False  True  True  True  True]
print("Sales > 150:", sales[mask_150])        # [180 160 210 175]
print("Days  > 150:", days[mask_150])         # Thursday Friday Saturday Sunday

mask_avg = sales >= sales.mean()
print("Days >= average:", days[mask_avg])     # Thursday Friday Saturday Sunday
print("Number of days > 150:", mask_150.sum())   # True counts as 1 -> 4
```

**Business answers**
- **4 days** exceeded 150.
- At or above the weekly average (155.71): **Thursday, Friday, Saturday, Sunday**.
- **Conclusion for the manager:** the second half of the week sells the most (Saturday is the best day with 210) while Wednesday is the weakest (95) → plan stock and staff for Thursday–Sunday and run a promotion on Wednesday.

## Bonus — what else NumPy can do with this data

```python
print("Std deviation:", sales.std().round(2))
print("Sorted:", np.sort(sales))
print("Sales +10 %:", sales * 1.10)                  # vectorised: applies to every value
print("Weekend total:", sales[5:].sum())             # Saturday + Sunday
week2 = np.array([130, 140, 100, 170, 165, 220, 180])
both = np.array([sales, week2])                      # 2D array: 2 rows x 7 columns
print(both.shape, both.sum(axis=1), both.mean(axis=0))   # total per week, average per day
```
