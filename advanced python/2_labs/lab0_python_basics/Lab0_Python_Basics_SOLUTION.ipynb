## Exercise 4 — Multiplication table + running total

```python
n = int(input("Enter a number: "))
total = 0
for i in range(1, 11):            # 1 .. 10 (11 is excluded)
    result = n * i
    print(f"{n} x {i} = {result}")
    total += result               # same as total = total + result
print("Sum of the table:", total) # n = 3 -> 165
```

## Exercise 9 — Nested lists: the board

```python
board = [["_", "_", "_"],
         ["_", "_", "_"],
         ["_", "_", "_"]]

def print_board(board):
    for row in board:
        print(" | ".join(row))

print_board(board)
board[1][1] = "X"        # row 1, column 1 = centre
print()
print_board(board)
```

## Exercise 10 — CHALLENGE: Tic-Tac-Toe (with the bonus winner check)

```python
board = [["_", "_", "_"],
         ["_", "_", "_"],
         ["_", "_", "_"]]

def print_board(board):
    for row in board:
        print(" | ".join(row))

def has_won(board, mark):
    for i in range(3):
        if board[i][0] == board[i][1] == board[i][2] == mark:   # row i
            return True
        if board[0][i] == board[1][i] == board[2][i] == mark:   # column i
            return True
    if board[0][0] == board[1][1] == board[2][2] == mark:       # diagonal
        return True
    if board[0][2] == board[1][1] == board[2][0] == mark:       # anti-diagonal
        return True
    return False

print("Welcome to Tic-Tac-Toe!")
print_board(board)
winner = None

for turn in range(9):
    player = "X" if turn % 2 == 0 else "O"          # 1. whose turn
    print(f"\nPlayer {player}'s turn")
    row = int(input("Row (0-2): "))                  # 2. ask row / column
    col = int(input("Column (0-2): "))

    if board[row][col] == "_":                       # 3. empty cell?
        board[row][col] = player
    else:
        print("That cell is already taken — turn skipped!")

    print_board(board)                               # 4. show the board
    if has_won(board, player):                       # bonus
        winner = player
        print(f"Player {player} wins!")
        break

if winner is None:
    print("Game over — it's a draw!")
```
