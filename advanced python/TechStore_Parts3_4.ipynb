### Questions

**1. What is the difference between a CSV file and a Pandas DataFrame?**
A CSV file is a plain text file stored on disk, where each line is a row and values are separated by commas. It has no types: everything is just text.
A DataFrame is a table in the computer's memory, created by pandas. It has rows, named columns, an index and a data type for each column, so we can filter, calculate and analyse the data.

**2. What does `pd.read_csv()` do?**
It reads a CSV file, splits each line into columns, guesses the type of each column (numbers, text…), treats empty cells as missing values (`NaN`) and returns the result as a DataFrame.

**3. What is the difference between `shape` and `size`?**
`shape` gives the dimensions as a tuple `(number of rows, number of columns)`.
`size` gives the total number of cells: rows × columns. For example, a shape of `(1003, 10)` gives a size of `10030`.

**4. What does the index represent in a DataFrame?**
The index is the label of each row. By default it is 0, 1, 2, … It is used to identify and select rows (for example with `.loc`), and pandas uses it to align rows when combining data.

**5. Why is `index=False` useful when exporting a CSV file?**
Without it, pandas writes the index as an extra unnamed column in the file. When the file is loaded again, this column appears as `Unnamed: 0`, which is useless and confusing. `index=False` keeps only the real data columns.

### Sales questions

```python
print("Transactions before cleaning:", len(sales))
print("Different customers:", sales["Customer_ID"].nunique())
print("Different products sold:", sales["Product_ID"].nunique())
print("Different regions:", sales["Region"].nunique())
print("Different sales representatives:", sales["Sales_Rep"].nunique())
print()
print("Average quantity purchased:", round(sales["Quantity"].mean(), 2))
print("Maximum unit price:", sales["Unit_Price"].max())
print("Minimum unit price:", sales["Unit_Price"].min())
print("Average discount:", round(sales["Discount"].mean(), 2))
```

`Order_Date` is read as text, so we convert it to a real date before looking for the oldest one.
`errors="coerce"` turns invalid or empty dates into `NaT` (missing date) instead of crashing.

```python
sales["Order_Date"] = pd.to_datetime(sales["Order_Date"], errors="coerce")
print("Oldest transaction date:", sales["Order_Date"].min())
```
