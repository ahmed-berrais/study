## Exercise 5 — Text cleaning and dates `Intermediate`

1. Print the distinct values of `city`. You should see typos (spaces, lower/upper case).
2. Clean `city` with `.str.strip()` and `.str.title()` and check that only **10 cities** remain.
3. Convert `order_date` to a real datetime with `pd.to_datetime(..., errors="coerce")`.
4. Drop the rows without a valid date and print the number of rows.

**Expected result:** Before: about `50` city spellings. After: `10` cities. `order_date` has dtype `datetime64` and `11,980` rows remain.

```python
print(orders["city"].nunique())
print(orders["city"].unique())

# clean city
orders["city"] = orders["city"].str.strip().str.title()
print(orders["city"].nunique())
print(orders["city"].unique())

# convert dates
orders["order_date"] = pd.to_datetime(orders["order_date"], errors="coerce")
print(orders["order_date"].dtype)

# drop rows without date
orders = orders.dropna(subset=["order_date"])
print(len(orders))
```

---

## Exercise 6 — Outliers and imputation `Challenge`

1. Look at `orders["quantity"].value_counts().sort_index()`. The business rule says **an order line never exceeds 10 units**. Remove the impossible rows.
2. The remaining missing `unit_price` values should be filled with the **median price of the same product** (hint: `groupby("product_id")["unit_price"].transform("median")`).
3. Convert `quantity` to integers (`int64`).
4. Check that there is no missing value left in `unit_price`, `quantity` and `order_date`.

**Expected result:** No `quantity` above 10, `0` missing prices, `quantity` dtype `int64`.

```python
print(orders["quantity"].value_counts().sort_index())

# 1. remove outliers (business rule: max 10 units)
orders = orders[orders["quantity"] <= 10]

# 2. impute unit_price with the median price of the same product
product_median = orders.groupby("product_id")["unit_price"].transform("median")
orders["unit_price"] = orders["unit_price"].fillna(product_median)

# 3. cast quantity
orders["quantity"] = orders["quantity"].astype("int64")

# 4. checks
print(orders[["unit_price", "quantity", "order_date"]].isna().sum())
print("Max quantity:", orders["quantity"].max())
print(orders.dtypes)
```

1. Save the clean data to `orders_clean_lab2.csv` (without the index).
2. Save the first 500 rows to `orders_sample.xlsx`.
3. Reload the CSV with `parse_dates=["order_date"]` and verify that the shape and the `order_date` dtype are the same.
4. Print a small **cleaning report**: rows in the raw file (12,060), rows in the clean file, and the share of rows removed.

**Expected result:** Same shape after reload; `order_date` is `datetime64`. Clean file: `11,965` rows — about **0.8 %** of the raw rows were removed (60 duplicates, 20 missing dates, 15 impossible quantities, minus overlaps).
